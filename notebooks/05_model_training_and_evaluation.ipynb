```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sys
import os

# Add src to path
sys.path.append('../src')

from preprocessor import Preprocessor
from modeling import ModelTrainer
from data_loader import DataLoader

import warnings
warnings.filterwarnings('ignore')

sns.set_style('whitegrid')
%matplotlib inline
```

## 1. Load Data

We will load both the credit card dataset and the engineered e-commerce fraud dataset.

```python
# Load datasets
creditcard_df = pd.read_csv('../data/processed/cleaned_creditcard.csv')
fraud_df = pd.read_csv('../data/processed/fraud_data_engineered.csv')

print(f"Credit Card Data Shape: {creditcard_df.shape}")
print(f"E-commerce Fraud Data Shape: {fraud_df.shape}")
```

## 2. Task 2a: Data Preparation and Baseline Model

### 2.1 Credit Card Dataset

We will perform a stratified train-test split to preserve the class distribution of the highly imbalanced credit card data.

```python
# Initialize Preprocessor and ModelTrainer
preprocessor = Preprocessor()
trainer = ModelTrainer(random_state=42)

# Prepare features and target
X_cc, y_cc = preprocessor.prepare_for_modeling(creditcard_df, target_col='Class', handle_imbalance=False)

print(f"Features shape: {X_cc.shape}")
print(f"Target distribution:\n{y_cc.value_counts(normalize=True)}")

# Stratified train-test split (80/20)
X_train_cc, X_test_cc, y_train_cc, y_test_cc = preprocessor.stratified_split(X_cc, y_cc, test_size=0.2)
```

#### Baseline Logistic Regression - Credit Card

```python
# Train baseline model
lr_cc = trainer.train_baseline_logistic_regression(X_train_cc, y_train_cc)

# Evaluate model
metrics_lr_cc = trainer.evaluate_model(lr_cc, X_test_cc, y_test_cc, 'LR_CreditCard')

# Plot Confusion Matrix
trainer.plot_confusion_matrix('LR_CreditCard')
```

### 2.2 E-commerce Fraud Dataset

Similar steps for the e-commerce dataset.

```python
# Prepare features and target
X_fraud, y_fraud = preprocessor.prepare_for_modeling(fraud_df, target_col='class', handle_imbalance=False)

print(f"Features shape: {X_fraud.shape}")
print(f"Target distribution:\n{y_fraud.value_counts(normalize=True)}")

# Stratified train-test split (80/20)
X_train_fraud, X_test_fraud, y_train_fraud, y_test_fraud = preprocessor.stratified_split(X_fraud, y_fraud, test_size=0.2)
```

#### Baseline Logistic Regression - E-commerce

```python
# Train baseline model
lr_fraud = trainer.train_baseline_logistic_regression(X_train_fraud, y_train_fraud)

# Evaluate model
metrics_lr_fraud = trainer.evaluate_model(lr_fraud, X_test_fraud, y_test_fraud, 'LR_Ecommerce')

# Plot Confusion Matrix
trainer.plot_confusion_matrix('LR_Ecommerce')
```
